```python
%sql
CREATE CATALOG IF NOT EXISTS corretora;
USE CATALOG corretora;

CREATE SCHEMA IF NOT EXISTS raw    COMMENT 'Arquivos de origem, imutáveis';
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Dados brutos em Delta, via Auto Loader';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'Dados limpos e tipados';
CREATE SCHEMA IF NOT EXISTS gold   COMMENT 'Modelo dimensional e métricas da corretora';
CREATE SCHEMA IF NOT EXISTS ops    COMMENT 'Controles operacionais da plataforma de dados';

CREATE VOLUME IF NOT EXISTS raw.landing;
CREATE VOLUME IF NOT EXISTS ops.checkpoints;

CREATE TABLE IF NOT EXISTS ops.log_ingestao (
  id_coleta         STRING    COMMENT 'Mesmo id gravado no envelope da Raw',
  fonte             STRING,
  entidade          STRING,
  url               STRING    COMMENT 'Sem token',
  status_code       INT,
  sucesso           BOOLEAN,
  latencia_ms       BIGINT,
  arquivo_raw       STRING,
  erro              STRING,
  executado_em_utc  TIMESTAMP
) COMMENT 'Uma linha por chamada de extração';
```
